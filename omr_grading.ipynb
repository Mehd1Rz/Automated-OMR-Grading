```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
import csv
from pathlib import Path


def get_results_dir():
    """<current working folder>/Results  - created automatically if missing."""
    folder = Path.cwd() / "Results"
    folder.mkdir(parents=True, exist_ok=True)
    return folder


class CalssScore():
    
    

    
    def init(self,path_of_aswer_sheet,path_of_key):
        self.path_of_aswer_sheet = path_of_aswer_sheet
        self.path_of_key = path_of_key    
        self.img = cv2.resize(cv2.imread(self.path_of_aswer_sheet),(1000,1000))




# تابع محاسبه ی درصد
    def Score2(self):
        self.Choises = []
        self.Key_dict = []
        
            



        img=cv2.imread(self.path_of_key)
        img_c1 = img[400:1300,280:1600]
        img_c2 = img[1350:2225,280:1600]
        img_c3 = img[2290:3150,280:1600]
        img_c4 = img[3230:4100,280:1600]
        img_c5 = img[4180:5040,280:1600]
        img_c6 = img[5120:5980,280:1600]
        img_c7 = img[6060:6920,280:1600]
        img_c8 = img[6990:7850,280:1600]
        img_c9 = img[7925:8780,280:1600]
        img_c10 = img[8860:9720,280:1600]
        img_c11 = img[9800:10650,280:1600]
        img_c12 = img[10730:11590,280:1600]
        img_c13 = img[11670:12530,280:1600]
        img_c14 = img[12610:13470,280:1600]
        img_c15 = img[13550:14410,280:1600]
        img_c16 = img[14490:15350,280:1600]
        img_c17 = img[15420:15840,280:1600]


        All_Box_1 = [img_c1,img_c2,img_c3,img_c4,img_c5,img_c6,img_c7,img_c8,img_c9,img_c10,img_c11,img_c12,img_c13,img_c14,img_c15,img_c16,img_c17]
        All_Box_2 = []

        for i in All_Box_1:
            Gray_of_Mask =  cv2.cvtColor(cv2.cvtColor(cv2.inRange(cv2.cvtColor(i, cv2.COLOR_BGR2HSV), np.array([38, 32, 25]), np.array([255, 255, 255])) , cv2.COLOR_GRAY2BGR),cv2.COLOR_BGR2GRAY)
            Resized_GrayMask = cv2.resize(Gray_of_Mask,(500,500))
            All_Box_2.append(Resized_GrayMask)
        Count_B = 1
        for j in All_Box_2:
            if Count_B < 17:
                vs = 10
            if Count_B > 16:
                vs = 5
            rows = np.vsplit(j,vs)
            for r in rows:
                cols = np.hsplit(r,4)
                A = cv2.countNonZero(cols[0])
                B = cv2.countNonZero(cols[1])
                C = cv2.countNonZero(cols[2])
                D = cv2.countNonZero(cols[3])
                if A > 0:
                    self.Key_dict.append(1)
                elif B > 0:
                    self.Key_dict.append(2)
                elif C > 0:
                    self.Key_dict.append(3)
                elif D > 0:
                    self.Key_dict.append(4)
                else:
                    self.Key_dict.append(0)
            Count_B+=1
        for i in range(0,5):
            self.Key_dict.append(0)




        def cut_pic(img):
            img1 = cv2.resize(img[265:385 , 84:165],(400,500))
            img2 = cv2.resize(img[395:515 , 84:165],(400,500))
            img3 = cv2.resize(img[527:650 , 84:165],(400,500))
            img4 = cv2.resize(img[660:775 , 84:165],(400,500))
            img5 = cv2.resize(img[790:905 , 84:165],(400,500))


            img6 = cv2.resize(img[265:385 , 240:320],(400,500))
            img7 = cv2.resize(img[395:515 , 240:320],(400,500))
            img8 = cv2.resize(img[527:650 , 240:320],(400,500))
            img9 = cv2.resize(img[660:775 , 240:320],(400,500))
            img10 = cv2.resize(img[790:905 , 240:320],(400,500))


            img11 = cv2.resize(img[265:385 , 395:480],(400,500))
            img12 = cv2.resize(img[395:515 , 395:480],(400,500))
            img13 = cv2.resize(img[525:650 , 395:480],(400,500))
            img14 = cv2.resize(img[660:775 , 397:480],(400,500))
            img15 = cv2.resize(img[790:905 , 395:480],(400,500))


            img16 = cv2.resize(img[265:385 , 555:635],(400,500))
            img17 = cv2.resize(img[395:515 , 555:635],(400,500))



            return img1,img2,img3,img4,img5,img6,img7,img8,img9,img10,img11,img12,img13,img14,img15,img16,img17

        pre_process_list = []
        def pre_process(img):
            img_G = cv2.cvtColor(img , cv2.COLOR_BGR2GRAY)
            ret,img_p = cv2.threshold(img_G, 127, 255, cv2.THRESH_BINARY_INV)
            pre_process_list.append(img_p)

        for cuted_pic in list(cut_pic(self.img)):
            pre_process(cuted_pic)


        def splitBoxes(img):
            boxes = []
            rows = np.vsplit(img , 10)
            for r in rows:
                cols = np.hsplit(r,4)
                A = cv2.countNonZero(cols[0])
                B = cv2.countNonZero(cols[1])
                C = cv2.countNonZero(cols[2])
                D = cv2.countNonZero(cols[3])
                rrr = 1230
                test_l = []
                if A >= rrr :
                    test_l.append(A)
                if  B>= rrr :
                    test_l.append(B)
                if C >= rrr :
                    test_l.append(C)
                if D >= rrr :
                    test_l.append(D)
                if len(test_l) == 0:
                    self.Choises.append(0)
                
                elif len(test_l) == 1:
                    if test_l[0] == A:
                        self.Choises.append(1)
                    elif test_l[0] == B:
                        self.Choises.append(2)
                    elif test_l[0] == C:
                        self.Choises.append(3)
                    elif test_l[0] == D:
                        self.Choises.append(4)
                elif len(test_l) != 1:
                    zzz = max(test_l)
                    if zzz == A:
                        self.Choises.append(1)
                    elif zzz == B:
                        self.Choises.append(2)
                    elif zzz == C:
                        self.Choises.append(3)
                    elif zzz == D:
                        self.Choises.append(4)
                        
                        
                        
                        
                        
                        


        for p_c_p in pre_process_list:
            splitBoxes(p_c_p)
        
        def Score_2(path_of_key):
            a = 0
            b = 5
            for i in range(0,170):
                if self.Choises[148] == 0:
                    b = 6

                if self.Choises[i] == self.Key_dict[i]:
                    a+=1
                else:
                    continue

            TP = a-b
            Final_Score = (TP / 164) * 100
            return Final_Score
        
        return round(Score_2(self.path_of_key),2) , self.Choises , self.Key_dict
    
    @staticmethod
    def Score(img_path,key_path):
        run_score = CalssScore()
        run_score.init(img_path,key_path)
        S = run_score.Score2()[0]
        return S

#تابع سیو کردن وضعیت برگه ی مورد نظر 
    def Save_Status(self,img_path,key_path):
        aa = 0
        bb = 5
        run_score = CalssScore()
        run_score.init(img_path,key_path)
        CH = run_score.Score2()[1]
        AN = run_score.Score2()[2]
        AN = AN[0:165]
        CH = CH[0:165]
        stat_list = []
        for i in range(0,165):
            if CH[i] == AN[i]:
                stat_list.append('True')
            if CH[i] != AN[i]:
                if CH[i] == 0:
                    stat_list.append('-')
                if CH[i] != 0:
                    stat_list.append('False')
            else:
                continue
        stat_list[148] = 'Not Count'
        
        Q_Num = []
        Final_List_for_CSV = []

        for ji in range(1,166):
            Q_Num.append(str(ji))

        for ki in range(0,165):
            O_1_list_for_Csv = []
            O_1_list_for_Csv.append(Q_Num[ki])
            O_1_list_for_Csv.append(stat_list[ki])
            Final_List_for_CSV.append(O_1_list_for_Csv)

        row = ['Qustion_Num','Status']
        out_path = get_results_dir() / (Path(img_path).stem + '.csv')
        with open(out_path, 'w', newline='') as stud:
            y = csv.writer(stud)
            y.writerow(row)
            y.writerows(Final_List_for_CSV)
        print('File Saved\nPath Of The File : ' + str(out_path))
    
# تابع نشان دادن سوالات به صورت بصری    
    def represention(self,img_path,key_path):
        run_score = CalssScore()
        run_score.init(img_path,key_path)
        CH = run_score.Score2()[1]
        CH = CH[0:165]
        AN = run_score.Score2()[2]
        AN = AN[0:165]
        Choise1 = []
        Choise2 = []
        Choise3 = []
        Choise4 = [] 
        for i1 in CH:
            if i1 == 1:
                Choise1.append('***')
            else:
                Choise1.append('---')
        for i1 in CH:
            if i1 == 2:
                Choise2.append('***')
            else:
                Choise2.append('---')
        for i1 in CH:
            if i1 == 3:
                Choise3.append('***')
            else:
                Choise3.append('---')
        for i1 in CH:
            if i1 == 4:
                Choise4.append('***')
            else:
                Choise4.append('---')    

        Status_Of_Q_List = []
        for jk in range(0,165):
            if CH[jk] == AN[jk]:
                Status_Of_Q_List.append('True')
            elif CH[jk] != AN[jk]:
                if CH[jk] == 0:
                    Status_Of_Q_List.append('Not Answerd')
                elif CH[jk] != 0:
                    Status_Of_Q_List.append('False')
        Status_Of_Q_List[148] = 'Deleted Qustion'
        Qustions_Numbers = []
        for i in range(1,166):
            Qustions_Numbers.append(i)
        df = pd.DataFrame(Qustions_Numbers,columns=['Qustions Numbers'])
        df['1'] = Choise1
        df['2'] = Choise2
        df['3'] = Choise3
        df['4'] = Choise4
        df['Status'] = Status_Of_Q_List
        df['Correct Answes'] = AN
        return df
    
#تابع مربوط به سیو کردن وضعیت هر سوال مریوط به تک تک برگه ها  
    @staticmethod

    def save_all_status(list_sheet , path_of_key):
        L_Main = list_sheet
        All_File_Name = [Path(z).stem for z in L_Main]
        ALL_STAT_LISTS = []
        for gg in L_Main:
            run_score = CalssScore()
            run_score.init(gg,path_of_key)
            CH = run_score.Score2()[1]
            AN = run_score.Score2()[2]
            AN = AN[0:165]
            CH = CH[0:165]
            stat_list = []

            for i in range(0,165):
                if CH[i] == AN[i]:
                    stat_list.append('True')
                elif CH[i] != AN[i]:
                    if CH[i] == 0:
                        stat_list.append('-')
                    elif CH[i] != 0:
                        stat_list.append('False')
            stat_list[148] = 'Not Count'
            ALL_STAT_LISTS.append(stat_list)
            QN = []
            for i in range(1,166):
                QN.append(i)
            df = pd.DataFrame(QN,columns=['Qustions Numbers'])
            for nums in range(0,len(ALL_STAT_LISTS)):
                df[All_File_Name[nums]] = ALL_STAT_LISTS[nums]

        out_path = get_results_dir() / 'Saved_All_Status.csv'
        df.to_csv(out_path, index=False)
        print('File Saved\nPath Of The File : ' + str(out_path))
    
# تابع مربوط به سیو تمام درصد ها
    @staticmethod
    def save_all(list_of_sheets,key_p):
        All_File_Name = [Path(z).stem for z in list_of_sheets]
        All_Scores = []
        for mems in list_of_sheets:
            run_score = CalssScore()
            All_Scores.append(run_score.Score(mems,key_p))
        
        df = pd.DataFrame(All_File_Name,columns=['File Name'])
        df['Score'] = All_Scores
        out_path = get_results_dir() / 'Saved_All_Scores.csv'
        df.to_csv(out_path, index=False)
        print('File Saved\nPath Of The File : ' + str(out_path))

        
        
#  این تابع تعداد درست های بزگه هارو بر میگردوند برا چک کردن       
#     @staticmethod
#     def show_T(img_path,key_path):
#         run_score = CalssScore()
#         run_score.init(img_path,key_path)
#         CH = run_score.Score2()[1]
#         CH = CH[0:165]
#         AN = run_score.Score2()[2]
#         AN = AN[0:165]
#         a = 3
#         Status_Of_Q_List = []
#         for jk in range(0,165):
#             if CH[jk] == AN[jk]:
#                 Status_Of_Q_List.append('True')
#             elif CH[jk] != AN[jk]:
#                 if CH[jk] == 0:
#                     Status_Of_Q_List.append('Not Answerd')
#                 elif CH[jk] != 0:
#                     Status_Of_Q_List.append('False')
#         Status_Of_Q_List[148] = 'Deleted Qustion'
#         for iu in Status_Of_Q_List:
#             if iu == 'True' :
#                 a+=1
#             else:
#                 continue
#         print(a)
```
